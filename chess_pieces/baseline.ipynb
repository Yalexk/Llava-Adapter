```python
# Load in the chess pieces dataset
# Simple toy dataset with 50 images to test learning capability
from datasets import load_dataset
import torch
from transformers import pipeline
from tqdm import tqdm

print(torch.backends.mps.is_available())

dataset = load_dataset('Trelis/chess_pieces')
print(dataset)
print(dataset["test"].features) 

from transformers import pipeline

pipe = pipeline("image-text-to-text", model="llava-hf/llava-1.5-7b-hf", device="mps", dtype=torch.float16)

def make_messages(img):
    return [{
        "role": "user",
        "content": [
            {"type": "image", "image": img},
            {"type": "text", "text": "Describe the chess piece/s within this image"},
        ],
    }]

def run(row):
    return pipe(text=make_messages(row["image"]), max_new_tokens=30,
                return_full_text=True)[0]["generated_text"]

run(dataset["test"][0])

for i in tqdm(range(len(dataset["test"]))):
    row = dataset["test"][i]
    tqdm.write(run(row))
```
